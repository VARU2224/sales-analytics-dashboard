```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print("All libraries imported successfully!")
```

```python

data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone",
                "Tablet", "Laptop", "Phone", "Tablet", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics", "Electronics",
                 "Electronics"],
    "Region": ["North", "South", "East", "West", "North",
               "South", "East", "West", "North", "South"],
    "Sales": [50000, 30000, 20000, 45000, 35000,
              22000, 55000, 32000, 18000, 60000],
    "Profit": [8000, 5000, 3000, 7000, 6000,
               2500, 10000, 4500, 2000, 12000],
    "Quantity": [5, 10, 8, 4, 12, 7, 6, 9, 5, 6]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print("Dataset created successfully!")
df.head()
```

```python

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nTotal Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())

df.describe()
```

```python

# Calculate total sales for each product
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

# Create the bar chart
plt.figure(figsize=(9, 5))
product_sales.plot(kind="bar")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

product_profit = df.groupby("Product")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
product_profit.plot(kind="bar")

plt.title("Total Profit by Product")
plt.xlabel("Product")
plt.ylabel("Profit (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

print("Sales by Region:")
print(region_sales)

print("\nBest Performing Region:", region_sales.idxmax())
print("Lowest Performing Region:", region_sales.idxmin())
```

```python

# Create a summary dashboard

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = len(df)
total_quantity = df["Quantity"].sum()

fig, axes = plt.subplots(2, 2, figsize=(14, 9))

# Chart 1: Sales by product
df.groupby("Product")["Sales"].sum().plot(
    kind="bar", ax=axes[0, 0], title="Sales by Product"
)
axes[0, 0].set_ylabel("Sales (₹)")
axes[0, 0].tick_params(axis="x", rotation=0)

# Chart 2: Profit by product
df.groupby("Product")["Profit"].sum().plot(
    kind="bar", ax=axes[0, 1], title="Profit by Product"
)
axes[0, 1].set_ylabel("Profit (₹)")
axes[0, 1].tick_params(axis="x", rotation=0)

# Chart 3: Sales by region
df.groupby("Region")["Sales"].sum().sort_values().plot(
    kind="barh", ax=axes[1, 0], title="Sales by Region"
)
axes[1, 0].set_xlabel("Sales (₹)")

# Chart 4: Quantity by product
df.groupby("Product")["Quantity"].sum().plot(
    kind="bar", ax=axes[1, 1], title="Quantity Sold by Product"
)
axes[1, 1].set_ylabel("Quantity")
axes[1, 1].tick_params(axis="x", rotation=0)

fig.suptitle(
    f"SALES ANALYTICS DASHBOARD | Sales: ₹{total_sales:,.0f} | "
    f"Profit: ₹{total_profit:,.0f}",
    fontsize=15, fontweight="bold"
)

plt.tight_layout()
plt.show()

print("Total Sales: ₹{:,.0f}".format(total_sales))
print("Total Profit: ₹{:,.0f}".format(total_profit))
print("Total Records:", total_orders)
print("Total Quantity Sold:", total_quantity)
```

```python

# Business Insights

product_sales = df.groupby("Product")["Sales"].sum()
product_profit = df.groupby("Product")["Profit"].sum()
region_sales = df.groupby("Region")["Sales"].sum()

best_product = product_sales.idxmax()
lowest_product = product_sales.idxmin()
most_profitable_product = product_profit.idxmax()
best_region = region_sales.idxmax()
lowest_region = region_sales.idxmin()

print("=" * 45)
print("       BUSINESS INSIGHTS REPORT")
print("=" * 45)

print(f"\n1. Total Revenue: Rs. {df['Sales'].sum():,}")
print(f"2. Total Profit: Rs. {df['Profit'].sum():,}")
print(f"3. Best-Selling Product: {best_product}")
print(f"4. Lowest-Selling Product: {lowest_product}")
print(f"5. Most Profitable Product: {most_profitable_product}")
print(f"6. Highest-Sales Region: {best_region}")
print(f"7. Lowest-Sales Region: {lowest_region}")

print("\nRECOMMENDATIONS")
print("-" * 45)
print(f"1. Analyze demand and consider promoting {best_product}.")
print(f"2. Investigate why {lowest_product} has lower sales.")
print(f"3. Study successful sales strategies in {best_region}.")
print(f"4. Investigate opportunities to improve sales in {lowest_region}.")
```
